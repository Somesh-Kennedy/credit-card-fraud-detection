# Exploratory Data Analysis

This notebook explores the transaction dataset, fraud distribution, feature distributions, and uses basic plots to highlight class imbalance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../credit_card_transactions.csv')
df.head()
```

```python
df['is_fraud'].value_counts().plot(kind='bar')
plt.title('Fraud class distribution')
plt.xlabel('is_fraud')
plt.ylabel('count')
plt.show()
```
